## 2. 变量、数据类型与算术运算

### 2.1 用变量保存数据

只把算式算出来还不够，我们通常需要把数据存下来留着后面用。在 Python 里，用等号 `=` 给数据起个名字，就叫做定义一个**变量**。

注意编程里的 `=` 不是数学上的“相等”，而是**赋值**——意思是“把右边算出来的结果，存到左边这个名字名下”。

下面我们用经典的 p 型有机半导体**并五苯（pentacene）**来举例：

````python
name = "pentacene"     # 字符串 (str)：文本内容，必须用单引号或双引号包起来
homo = -5.0            # 浮点数 (float)：带小数点的数，这里表示 HOMO 能级 (eV)
lumo = -3.2            # 浮点数 (float)：LUMO 能级 (eV)
n_rings = 5            # 整数 (int)：并五苯由 5 个苯环并在一起组成
is_p_type = True       # 布尔值 (bool)：逻辑真假，只有 True 和 False 两个值（首字母大写）

print(name, homo, lumo, n_rings, is_p_type)
````

如果不确定某个变量里存的是什么类型的数据，可以用内置函数 `type()` 查看：

````python
print(type(name))
print(type(homo))
print(type(n_rings))
print(type(is_p_type))
````

上面这四种（`str`、`float`、`int`、`bool`）是 Python 里最基础的单值类型。给变量起名字时，习惯用清晰的英文单词加下划线（比如 `n_rings`、`band_gap`），注意两点：
1. 变量名只能包含字母、数字和下划线 `_`，且**不能以数字开头**（写 `3ht = ...` 会报错，可以写 `mol_3ht`）。
2. **严格区分大小写**：`homo` 和 `HOMO` 在 Python 看来是两个毫无关系的变量。

### 2.2 算术运算

定义好变量之后，就可以直接拿变量名来做数学计算。先看一个最直接的例子——用刚才定义的 `lumo` 和 `homo` 计算并五苯的能隙（Band Gap）：

````python
gap = lumo - homo
print(gap)
````

你可能会注意到，`-3.2 - (-5.0)` 打印出来居然是 `1.7999999999999998`，而不是干干净净的 `1.8`。这不是你的代码写错了，而是因为计算机底层用二进制存储小数，绝大多数十进制小数转换成二进制时都会有一点点截断误差（在大概小数点后第 16 位）。平时计算完全不用管它，如果希望打印出来好看，可以用 `round(gap, 2)` 四舍五入保留两位小数。

除了加减乘除，Python 还支持乘方、整除和取余运算：

````python
wavelength = 1240 / gap     # 除法 /：把能隙 (eV) 换算成吸收边波长 (nm)，结果永远是浮点数
print("吸收边波长 (nm):", round(wavelength, 1))

# 并苯 (acene) 通式是 C_(4n+2)H_(2n+4)，用 n_rings = 5 算一下并五苯的碳原子数：
n_carbons = 4 * n_rings + 2
print("碳原子数:", n_carbons)

# 其他常用算术运算符：
print(2 ** 10)   # 乘方 **：2 的 10 次方（注意不是 ^，在 Python 里 ^ 是按位异或）
print(17 // 5)   # 整除 //：只保留商的整数部分，结果是 3
print(17 % 5)    # 取余 %：17 除以 5 的余数，结果是 2
````

有时我们需要在变量原有数值的基础上加减，比如统计总原子数时加上氢原子。可以写成 `n_atoms = n_atoms + 14`，也可以用简写 `+=`：

````python
n_atoms = n_carbons     # 先算上 22 个碳原子
n_atoms += 14           # 等价于 n_atoms = n_atoms + 14（加上 14 个氢原子）
print("并五苯总原子数:", n_atoms)
````

## 3. 字符串与格式化输出

### 3.1 用 f-string 拼接文字与数值

刚才我们算出了 `homo`、`lumo` 和 `gap`，如果想像写实验报告一样输出一句完整的话，用逗号一个个隔开 `print` 会显得很零碎。Python 里最舒服的做法是用 **f-string**：在引号前面加个字母 `f`，然后在字符串里直接用花括号 `{变量名}` 把变量填进去：

````python
print(f"{name} 的 HOMO = {homo} eV，LUMO = {lumo} eV，能隙约为 {gap:.2f} eV")
````

花括号里的 `:.2f` 是格式说明符，意思是“按浮点数（float）格式显示，并保留小数点后 2 位”。这样连 `round()` 都省了，后面打印模型误差、画图写标题时几乎天天都会用到。

### 3.2 字符串的下标、切片与常用操作

在化学和材料信息学里，字符串除了用来存名称，最重要的用途是表示分子结构——也就是 **SMILES**（用一行字符把分子的原子和成键方式写出来，第 `01d` 节会专门讲）。比如导电高分子 P3HT 的基本单元是**噻吩（thiophene）**，它的 SMILES 写作 `"c1ccsc1"`（小写字母代表芳香环上的原子，`c` 是碳，`s` 是硫）。

我们可以把字符串看作一排按顺序排好的字符，用方括号 `[]` 按位置提取其中的部分：

````python
thiophene = "c1ccsc1"

print("字符串长度:", len(thiophene))   # len() 返回一共有几个字符
print("第 1 个字符:", thiophene[0])    # 注意：Python 从 0 开始数！下标 0 才是第一个
print("最后 1 个字符:", thiophene[-1]) # 负数下标表示倒着数，-1 就是倒数第一个
print("切片 [1:4]:", thiophene[1:4])   # 取出下标 1、2、3 的字符（包含起点 1，不包含终点 4）
````

> **记住这两条铁律**：
> 1. Python 的编号（下标 / Index）永远**从 0 开始**。
> 2. 切片 `[a:b]` 永远是**左闭右开**——包含位置 `a`，但不包含位置 `b`。
>
> 后面学列表、NumPy 数组和 Pandas 表格时，全都是同一套规则。

除了按位置取字符，字符串还自带很多实用的操作工具（写在字符串变量后面，加个点 `.` 调用）：

````python
hexyl = "CCCCCC"                     # 正己基侧链的 SMILES

print(hexyl + thiophene)             # 用 + 把两个字符串首尾拼起来（3-己基噻吩）
print("s" in thiophene)              # 用 in 检查里面包不包含某个子串（是否含硫原子）
print(thiophene.count("c"))          # .count() 数一数出现了几个芳香碳 'c'
print(thiophene.upper())             # .upper() 全部转成大写

# 读实验数据文件时最常用的两个清洗操作：.strip() 和 .split()
raw_line = "  pentacene, -5.0, -3.2 \n"
clean_line = raw_line.strip()        # 去掉首尾多余的空格和换行符 \n
print(clean_line.split(","))         # 按逗号切开，变成一串独立的字段
````

### 4.1 列表（`list`）：按顺序排列的一组数据

列表用方括号 `[]` 表示，里面的元素用逗号隔开。它就像一列编好号的格子，既可以按下标取值，也可以随时往里添减内容：

````python
# 并苯系列：苯、萘、蒽、并四苯
acenes = ["benzene", "naphthalene", "anthracene", "tetracene"]

print(acenes[0])              # 取第一个：benzene
print(acenes[-1])             # 取最后一个：tetracene
print(acenes[1:3])            # 切片：取下标 1 和 2，得到子列表
print(len(acenes))            # 列表里有几个元素
print("pentacene" in acenes)  # 检查并五苯在不在列表里
````

刚才检查发现 `"pentacene"` 还不在列表里。我们可以用 `.append()` 把它追加到列表末尾，或者直接修改某个位置的元素：

````python
acenes.append("pentacene")    # 在列表末尾追加一个新元素
print("追加后:", acenes)

acenes[0] = "Benzene"         # 直接修改第 0 个位置的元素
acenes[0] = "benzene"         # 改回来
````

如果列表里装的都是数字（比如同一批次 5 个器件测出的载流子迁移率），可以直接用内置函数求最大值、最小值、总和，或者用 `sorted()` 排序：

````python
mobilities = [0.1, 1.0, 0.05, 20.0, 3.5]   # 迁移率 cm²/(V·s)，示意值

print("最高:", max(mobilities), "最低:", min(mobilities))
print("平均值:", sum(mobilities) / len(mobilities))
print("从小到大:", sorted(mobilities))
print("从大到小:", sorted(mobilities, reverse=True))
````

### 4.2 元组（`tuple`）：不可修改的固定组合

元组用圆括号 `()` 表示。它和列表几乎一样，唯一的区别是**一旦创建就不能修改**（不能 `.append()`，也不能改里面的值）。因此它常用来存放一组固定搭配的数据，比如晶胞的三个晶格常数 $(a, b, c)$、原子的三维坐标 $(x, y, z)$，或者神经网络里矩阵的维度。

元组还有一个很顺手的写法叫**拆包（unpacking）**，可以把里面的值一次性赋给多个变量：

````python
lattice = (5.9, 7.6, 15.6)   # 并五苯薄膜相晶格参数 a, b, c (Å)

a, b, c = lattice            # 拆包：把三个数分别赋给 a, b, c
print(f"a = {a} Å, b = {b} Å, c = {c} Å")
````

### 4.3 字典（`dict`）：按“名称”查找的键值对

如果我们要记录一个分子的好几项信息——名称、SMILES、HOMO、LUMO、导电类型——用列表装成 `["pentacene", -5.0, -3.2, "p"]` 会很不直观，过两天自己都忘了 `-5.0` 和 `-3.2` 哪个是 HOMO。

**字典**就是为了解决这个问题：它用花括号 `{}` 表示，里面存的是一对对的 **`键: 值`（`key: value`）**。查找数据时不用记位置下标，直接报名字（键）就行：

````python
pentacene = {
    "name": "pentacene",
    "smiles": "c1ccc2cc3cc4cc5ccccc5cc4cc3cc2c1",
    "homo": -5.0,
    "lumo": -3.2,
    "type": "p",
}

print("HOMO 能级:", pentacene["homo"])       # 用方括号 + 键名取值
print("有迁移率数据吗?", "mobility" in pentacene)

# 新增或修改条目，语法完全一样：
pentacene["mobility"] = 1.0                  # 原来没有 "mobility"，就新增一项
pentacene["homo"] = -4.9                     # 原来有 "homo"，就更新它的值

print("当前包含的键:", list(pentacene.keys()))
print(pentacene)
````

把**列表**和**字典**结合起来——每个分子是一个字典，再把好几个分子装进一个列表里——这就构成了一张微型的数据表（下一节 `01b` 要学的 `pandas.DataFrame` 本质上就是它的强化版）：

````python
molecules = [
    {"name": "pentacene", "homo": -5.0, "lumo": -3.2},
    {"name": "rubrene",   "homo": -5.4, "lumo": -3.1},
    {"name": "C60",       "homo": -6.1, "lumo": -4.0},
]

# 先用 [0] 取出第一个分子的字典，再用 ["name"] 取出它的名字
print(molecules[0]["name"], "的 HOMO 是", molecules[0]["homo"], "eV")
````

## 5. 条件判断（`if`）

写程序筛选材料时，我们经常要设定门槛：比如“能隙小于 2.0 eV **而且** 迁移率大于 0.5 cm²/(V·s)”。

### 5.1 比较与逻辑运算

我们先看看怎么用 Python 写出判断条件。比较两个数的大小或是否相等，得到的结果永远是布尔值（`True` 或 `False`）：

## 6. 循环与列表推导式

### 6.1 `for` 循环：把容器里的元素挨个处理一遍

在练习 01a-1 里，我们手动取出了 `molecules[1]` 来算能隙。如果列表里有成百上千个分子，总不能手写一千行代码。`for` 循环会依次把列表里的每一个元素取出来，交给循环体里的代码去处理：

````python
for mol in molecules:
    gap = mol["lumo"] - mol["homo"]
    print(f"{mol['name']:10s}  gap = {gap:.2f} eV")   # :10s 表示字符串占 10 个字符宽度，方便对齐
````

写 `for` 循环时，有三个内置小工具经常配合出场：
- `range(n)`：生成从 `0` 到 `n-1` 的整数序列，适合“把某件事重复做 $n$ 次”。
- `enumerate(列表)`：遍历列表时，同时把**元素下标**和**元素本身**一起拿出来。
- `zip(列表1, 列表2)`：像拉链一样，把两个等长的列表一对一配对遍历。

````python
n_rings_list = [1, 2, 3, 4, 5]

print("--- 使用 enumerate 同时获取序号和名称 ---")
for i, name in enumerate(acenes):
    print(f"第 {i} 号: {name}")

print("\n--- 使用 zip 配对遍历两个列表 ---")
for name, n in zip(acenes, n_rings_list):
    print(f"{name} 含有 {n} 个苯环")
````

### 6.2 列表推导式（List Comprehension）

数据处理中最常见的套路是：“遍历一个列表，把每个元素算一下，放进一个新列表里”。用普通的 `for` 循环写出来需要四行，而 Python 提供了一种浓缩成一行的优雅写法，叫做**列表推导式**：

````python
# 传统写法：先建空列表，再循环 append
gaps = []
for mol in molecules:
    gaps.append(round(mol["lumo"] - mol["homo"], 2))
print("传统循环结果:", gaps)

# 列表推导式：一行搞定，逻辑完全等价
gaps = [round(mol["lumo"] - mol["homo"], 2) for mol in molecules]
print("列表推导式结果:", gaps)

# 还可以在末尾加 if 条件顺手做筛选：只挑出能隙小于 2.2 eV 的分子名称
low_gap_mols = [mol["name"] for mol in molecules if (mol["lumo"] - mol["homo"]) < 2.2]
print("窄带隙分子:", low_gap_mols)
````

### 6.3 `while` 循环：按条件重复直到达标

`for` 循环适合事先知道要遍历哪些元素的情况；而有时候我们不知道要循环多少步，只知道“只要还没达到目标条件，就继续往下算”（比如量子化学里的结构弛豫、神经网络的迭代训练）。这时候就用 `while` 循环。

下面用一个简单的例子示意：从苯环（`n = 1`）开始逐个并环，假设每增加一个环能隙大约降低 0.75 eV，看看要并到几个环能隙才会降到 2.0 eV 以下：

````python
n = 1
gap = 4.8   # 单苯环的示意能隙 (eV)

while gap >= 2.0:
    print(f"{n} 个环时，gap ≈ {gap:.2f} eV（仍 >= 2.0 eV，继续并环）")
    n += 1
    gap -= 0.75

print(f"--> 到 {n} 个环时，gap 降到了 {gap:.2f} eV，循环结束。")
````

> 写 `while` 循环时一定要确保循环体里有更新变量的语句（如上面的 `gap -= 0.75`）。如果条件永远为 `True`，程序就会陷入停不下来的“死循环”——万一遇到了，点击工具栏上的 **■（停止 / Interrupt）** 按钮即可强行终止。

## 7. 函数（`def`）

回顾一下从第 2 节到现在，算能隙 `lumo - homo` 和换算波长 `1240 / gap` 的公式我们已经重复写了好几遍。如果某天你想把常数 `1240` 改成更精确的 `1239.84`，难道要翻遍整个脚本去逐个修改吗？

更好的办法是把这段计算逻辑打包成一个**函数（Function）**：用 `def` 给它起个名字、指定需要输入的**参数**，最后用 `return` 把算好的结果交出来。以后不管在哪里用，直接喊它的名字就行：

````python
def gap_to_wavelength(gap_ev):
    """将能隙 (eV) 换算为光吸收边波长 (nm)。"""
    wavelength_nm = 1239.84 / gap_ev
    return wavelength_nm

# 定义好之后，可以拿不同的数值反复调用：
print(round(gap_to_wavelength(1.8), 1))
print(round(gap_to_wavelength(3.0), 1))
````

函数下面第一行用三引号 `"""..."""` 写的文字叫做文档字符串（docstring），用来简要说明这个函数的用途和参数单位。

函数可以接收多个参数，还可以给参数设定**默认值**（调用时不传该参数，就自动使用默认值）：

````python
def summarize_molecule(name, homo, lumo, unit="eV"):
    gap = lumo - homo
    return f"{name}: HOMO = {homo} {unit}, LUMO = {lumo} {unit}, gap = {gap:.2f} {unit}"

# 按位置顺序传入参数（unit 不传，默认就是 "eV"）
print(summarize_molecule("pentacene", -5.0, -3.2))

# 也可以明确写出“参数名=值”，这样顺序打乱也没关系
print(summarize_molecule(homo=-6.1, lumo=-4.0, name="C60"))
````

````python
import numpy as np

homos = np.array([-5.0, -5.4, -6.1])
lumos = np.array([-3.2, -3.1, -4.0])

gaps = lumos - homos                 # 不用写循环，直接对应位置相减！
print("三个分子的能隙:", gaps)
print("对应的吸收波长:", 1240 / gaps)  # 每个元素同时被 1240 除

# 数组自带常用的统计方法，以及不带括号的属性：
print("均值:", gaps.mean(), "标准差:", round(gaps.std(), 3))
print("数组形状 (shape):", gaps.shape)

# 还可以直接把条件塞进方括号里做筛选（布尔索引）：
print("小于 2.2 eV 的能隙:", gaps[gaps < 2.2])
````

> **顺便理清一个常见困惑：为什么有的后面带括号 `()`，有的不带？**
> 在读上面的代码时，你会看到三种写法：
> 1. `len(gaps)`、`round(x, 2)`：独立的**函数**，把数据塞进括号里处理。
> 2. `gaps.mean()`、`acenes.append(...)`：对象自带的**方法（Method）**，本质上也是函数（要执行一个动作或计算），所以末尾必须带括号 `()`。
> 3. `gaps.shape`、`math.pi`：对象自带的**属性（Attribute）**，它只是一个现成存好的数值或尺寸信息，不需要执行计算动作，所以**不带括号**。下一节用 Pandas 时你会经常看到 `df.shape`（看表格几行几列，不带括号）和 `df.head()`（取前几行，带括号）。
